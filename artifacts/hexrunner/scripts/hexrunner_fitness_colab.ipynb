Upload `train_fitness_model.py` and the committed `fitnessWeights.json` when prompted. Run every cell, retain the visible output, then use **File → Download → Download .ipynb** so the execution evidence is preserved.

```python
from google.colab import files

print("Upload train_fitness_model.py and the committed fitnessWeights.json")
uploaded = files.upload()
required = {"train_fitness_model.py", "fitnessWeights.json"}
missing = required.difference(uploaded)
assert not missing, f"Missing required files: {sorted(missing)}"
print("Ready:", ", ".join(sorted(required)))
```

```python
import subprocess

result = subprocess.run(
    ["python", "train_fitness_model.py", "--output", "colabFitnessWeights.json"],
    check=True,
    capture_output=True,
    text=True,
)
print(result.stdout)
```

```python
import hashlib
import json
from pathlib import Path

committed = Path("fitnessWeights.json").read_bytes()
exported = Path("colabFitnessWeights.json").read_bytes()
model = json.loads(exported)

print("Provenance rows:", model["provenance"]["rows"])
print("Provenance epochs:", model["provenance"]["epochs"])
print("Committed SHA-256:", hashlib.sha256(committed).hexdigest())
print("Exported SHA-256: ", hashlib.sha256(exported).hexdigest())
print("BYTE-FOR-BYTE MATCH:", committed == exported)
assert model["provenance"]["rows"] == 500
assert model["provenance"]["epochs"] == 2000
assert committed == exported, "Export differs from the committed model"
print("PASS: 500 samples, 2,000 epochs, and exported weights verified.")
```
